```python
from transformers import AutoModel, AutoTokenizer
from transformers import AutoModelForTokenClassification
from safetensors.torch import load_file
import torch

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert_tokenizer_uncased_run7")

# Load the model from safetensors
model_path = "bert_model_uncased_run7"
model = AutoModelForTokenClassification.from_pretrained(model_path)


```

```python
from transformers import AutoTokenizer
import torch

sample_text = "starte einen anruf"  # replace with an example text

# Tokenize and prepare the input
inputs = tokenizer(sample_text)
inputs
```

```python
import onnx
# Define the path where the ONNX model will be saved
onnx_model_path = "bert_model_trained.onnx"

# Export to ONNX
torch.onnx.export(
    model,                               # The model to be exported
    (inputs["input_ids"],                # Example input tuple (input_ids only)
     inputs["attention_mask"]),          # Include attention mask if needed
    onnx_model_path,                     # Path to save the model
    input_names=["input_ids", "attention_mask"],  # Model input names
    output_names=["output"],             # Model output name
    dynamic_axes={                       # Set dynamic axes for variable length
        "input_ids": {0: "batch_size", 1: "sequence_length"},
        "attention_mask": {0: "batch_size", 1: "sequence_length"},
        "output": {0: "batch_size", 1: "sequence_length"}
    },
    opset_version=14                    # Set the ONNX opset version (11 or higher)
)
print(f"Model exported to {onnx_model_path}")

# Load the model
onnx_model = onnx.load(onnx_model_path)

# Check the model
onnx.checker.check_model(onnx_model)
print("ONNX model is valid!")
```

```python
import onnxruntime as ort
import numpy as np

onnx_model_path = "bert_model_trained.onnx"

# Load the ONNX model
session = ort.InferenceSession(onnx_model_path)

# Define the input data
# Replace these with your actual input values
input_ids = np.array([[102, 2629, 30937, 366, 14860, 103]])  # Example token IDs
attention_mask = np.array([[1, 1, 1, 1, 1, 1]])             # Attention mask

# Run inference
inputs = {
    "input_ids": input_ids,
    "attention_mask": attention_mask
}

# Get predictions
outputs = session.run(None, inputs)

# Display the output
print(outputs)  # Outputs will be a list with predictions for each token
```

```python
from scipy.special import softmax

# Assuming `outputs[0]` contains logits for each token
logits = outputs[0]
probabilities = softmax(logits, axis=-1)

# Get the predicted class for each token by taking the argmax across classes
predicted_labels = np.argmax(probabilities, axis=-1)
print(predicted_labels)
```

```python
label_names = ['O',
 'B-Action',
 'I-Action',
 'B-Entity',
 'I-Entity',
 'B-Person',
 'I-Person',
 'B-Task',
 'I-Task',
 'B-Time',
 'I-Time',
 'B-Distance',
 'B-Question',
 'I-Question']
id2label = {k: v for k, v in enumerate(label_names)}

def convert_labels_to_names(predictions, id2label):
    # Remove the first and last label
    predictions = predictions[1:-1]
    label_names = []
    for label_id in predictions:
        label_names.append(id2label.get(label_id, 'Unknown'))  # Fallback to 'Unknown' if label_id not found
    return label_names



# Run inference and convert output labels
label_names = [convert_labels_to_names(pred, id2label) for pred in predicted_labels]

print(label_names)
```

```python
model = onnx.load('bert_model_trained.onnx')

# Get the input and output names
input_names = [input.name for input in model.graph.input]
output_names = [output.name for output in model.graph.output]

print("Input Names:", input_names)
print("Output Names:", output_names)
```
