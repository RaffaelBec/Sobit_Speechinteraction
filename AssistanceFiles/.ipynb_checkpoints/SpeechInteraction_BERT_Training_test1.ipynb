```python
import re
from datasets import load_dataset

# Load the CSV file
dataset = load_dataset("csv", data_files=r"C:\5EHIF\Diplomprojekt\Sobit_Speechinteraction\AssistanceFiles\Resources\Traindata_processed_asCSV.csv", delimiter=";")

# Convert tokens and tags fields from strings to lists
def parse_fields(example):
    # Split tokens on spaces to get a list of words
    example["tokens"] = example["tokens"].split()

    # Parse tags using regex to remove brackets and commas
    tags_string = example["tags"]
    tags_list = re.findall(r'\b\w+[-]?\w*\b', tags_string)  # Capture each tag without brackets or commas
    example["tags"] = tags_list
    return example

# Apply parsing
dataset = dataset.map(parse_fields)

# Get unique label names from tags

print("Parsed Dataset:", dataset)
print(dataset["train"][0])
```

```python
label_names = set(label for example in dataset["train"]["tags"] for label in example)
label_names
```

```python
# Check for consistency
def check_consistency(example):
    if len(example["tokens"]) != len(example["tags"]):
        print("Inconsistent row found:", example)
    return example

# Apply the check to the dataset
dataset = dataset.map(check_consistency)

# View the processed dataset
print(dataset)
```

```python
# use autotokenizer
from transformers import AutoTokenizer
from transformers import AutoModelForTokenClassification, TrainingArguments, Trainer

# choose from multiple different tokenizers
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
#tokenizer = AutoTokenizer.from_pretrained("bert-base-multilingual-cased")


#model = AutoModelForTokenClassification.from_pretrained(
#    "distilbert/distilbert-base-uncased", num_labels=13, id2label=id2label, label2id=label2id)
```

```python
label2id = {label: idx for idx, label in enumerate(label_names)}
id2label = {idx: label for label, idx in label2id.items()}

def tokenize_adjust_labels(all_samples_per_split):
    # Tokenize and split into words
    tokenized_samples = tokenizer(
        all_samples_per_split["tokens"], 
        is_split_into_words=True, 
        padding=True, 
        truncation=True, 
        return_tensors="np"  # Use NumPy arrays for batched processing
    )
    
    # List to store adjusted labels for each sentence
    total_adjusted_labels = []

    for i in range(len(tokenized_samples["input_ids"])):
        word_ids = tokenized_samples.word_ids(batch_index=i)
        existing_labels = [label2id[label] for label in all_samples_per_split["tags"][i]]
        
        adjusted_labels = []
        current_word = -1  # Track word ID

        # Adjust labels to match tokenized output
        for word_id in word_ids:
            if word_id is None:
                adjusted_labels.append(-100)  # Mask special tokens
            elif word_id != current_word:
                adjusted_labels.append(existing_labels[word_id])  # Map label to first token in each word
                current_word = word_id
            else:
                adjusted_labels.append(existing_labels[word_id])  # Repeat label for subtokens

        total_adjusted_labels.append(adjusted_labels)
    
    # Add adjusted labels to tokenized samples
    tokenized_samples["labels"] = total_adjusted_labels
    return tokenized_samples

# Apply the function to the dataset
tokenized_dataset = dataset.map(tokenize_adjust_labels, batched=True)

# Inspect the tokenized dataset
print(tokenized_dataset)
```

Explanation of Changes
Tokenizer Initialization: Initialize AutoTokenizer for your specific model.
Label Mapping: Use label2id to convert labels from strings to IDs for easier processing.
Tokenization: tokenizer(..., return_tensors="np") returns NumPy arrays, which are efficient for batch processing.
Label Adjustment: The labels are adjusted to match the subword token structure, with -100 masking special tokens (used by the Trainer API for token classification).
This function now processes the tokens and tags fields from your dataset, prepares input_ids, attention_mask, and labels, and makes it compatible with the Trainer API.

```python
### Trainer API
from transformers import AutoModelForTokenClassification, TrainingArguments, Trainer
import numpy as np




def compute_metrics(p):
    predictions, labels = p
    predictions = np.argmax(predictions, axis=2)

    # Remove ignored index (special tokens)
    true_predictions = [
        [label_names[p] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]
    true_labels = [
        [label_names[l] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]

    results = metric.compute(predictions=true_predictions, references=true_labels)
    flattened_results = {
        "overall_precision": results["overall_precision"],
        "overall_recall": results["overall_recall"],
        "overall_f1": results["overall_f1"],
        "overall_accuracy": results["overall_accuracy"],
    }
    for k in results.keys():
      if(k not in flattened_results.keys()):
        flattened_results[k+"_f1"]=results[k]["f1"]

    return flattened_results
```

```python
from datasets import DatasetDict

# Assuming `dataset` is a DatasetDict with only a 'train' split
train_test_split = tokenized_dataset["train"].train_test_split(test_size=0.1)  # Use 10% for validation

# Create a new DatasetDict with train and validation splits
tokenized_dataset = DatasetDict({
    'train': train_test_split['train'],
    'validation': train_test_split['test']
})
```

### Model training / fine tuning
Next, we load the chosen model, set our training parameters and use the Trainer to fine tune the model.

All metrics will be locked with wandb. 
For runtime-memory usage, look at WandB. This works for training, fine-tuning and normal runtime usage.

```python
model = AutoModelForTokenClassification.from_pretrained("bert-base-multilingual-cased", num_labels=len(label_names))
training_args = TrainingArguments(
    output_dir="./fine_tune_bert_output",
    evaluation_strategy="steps",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=7,
    weight_decay=0.01,
    logging_steps = 1000,
    report_to="wandb",
    run_name = "ep_10_tokenized_11",
    save_strategy='no'     # no, epochs or steps
)
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=["train"],
    eval_dataset=tokenized_dataset["validation"],
    data_collator=data_collator,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics
)

# Save model
trainer.save_model("model")

trainer.train()
```
