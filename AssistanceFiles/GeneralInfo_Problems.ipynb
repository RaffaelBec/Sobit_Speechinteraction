### How to train / fine tune a model

1. Select the model -> what is the task at hand? Which models are available?
2. Select / Gather data
    - Data preprocessing (exclude NaNs, exclude outliers, remove stopwords)
    - Label data (Textinformation: set a class e.g. "Mayer" = [0 (Person)]
3. Train / Fine tune model with data
4. Evaluate model
5. If the model needs further fine tuning, proceed from step 2.

To train and test a model, the data is split into training, test and validation datasets e.g. (80%,15%,5%)- or use k-fold to determine the best training test split.

Learning rate:
    Learning rate scales the magnitute of the weight updates to minimize the loss function:

### Possible problems

1. Overfitting
The model is tune towards the training data, so much so that it is only able to correctly predict the training data. Any new "unseen" data cannot be predicted correctly, only if it is very similar to the training data.

2. Underfitting
The model was trained on only a few examples of data. Therefore it might correctly predict the new data, however a lot of missclassifications might happen.

Epochs: Number of full training iterations

Evaluation strategy:
The evaluation strategy to adopt during training. Possible values are:

        "no": No evaluation is done during training.

        "steps": Evaluation is done (and logged) every eval_steps.

        "epoch": Evaluation is done at the end of each epoch.
