## Convert model for Android

This notebook holds resources on how to convert a model to tensorflow lite / ONNX, so that it can be used within an Android application. The tutorials also show how such a model can be included within the Android application.

Models can be used in two different ways with Android either use the ONNX format directly from huggingfaces, or convert your model to TensorflowLite.
